```python
import pandas as pd
from transformers import pipeline

# Pin the model
classifier = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english",
)

#analysing customer reviews
reviews = [
    "The onboarding was smooth and support replied within minutes.",
    "App keeps crashing after the latest update. Very disappointed.",
    "It's okay, does the job but nothing special.",

]


results = classifier(reviews)
df = pd.DataFrame({
    "review": reviews,
    "sentiment": [r["label"] for r in results],
    "confidence": [round(r["score"], 3) for r in results],
})

display(df)
print(df["sentiment"].value_counts())
```

```python
import pandas as pd
from transformers import pipeline

clf = pipeline("sentiment-analysis",
               model="distilbert-base-uncased-finetuned-sst-2-english")
```

```python
tests = [
    ("easy",     "I love this phone, the camera is stunning.", "POSITIVE"),
    ("easy",     "Worst purchase ever. Total waste of money.", "NEGATIVE"),
    ("negation", "This is not bad at all.", "POSITIVE"),
    ("negation", "I can't say I'm happy with the service.", "NEGATIVE"),
    ("sarcasm",  "Oh great, another update that breaks everything.", "NEGATIVE"),
    ("sarcasm",  "Wow, thanks for making me wait 3 hours.", "NEGATIVE"),
    ("mixed",    "Great screen, terrible battery life.", "NEGATIVE"),
    ("slang",    "ngl kinda mid tbh", "NEGATIVE"),
    ("slang",    "this movie was fire 🔥🔥", "POSITIVE"),
]
df = pd.DataFrame(tests, columns=["category", "text", "expected"])
preds = clf(df.text.tolist())
df["predicted"] = [p["label"] for p in preds]
df["confidence"] = [round(p["score"], 3) for p in preds]
df["correct"] = df.expected == df.predicted

display(df)
print(df.groupby("category").correct.mean())
```

```python
models = {
    "distilbert-sst2 (2 classes)": "distilbert-base-uncased-finetuned-sst-2-english",
    "twitter-roberta (3 classes)": "cardiffnlp/twitter-roberta-base-sentiment-latest",
}
samples = ["The meeting is at 3 PM on Tuesday.",
           "ngl kinda mid tbh",
           "Great screen, terrible battery life."]

for name, m in models.items():
    p = pipeline("sentiment-analysis", model=m)
    print(name)
    for s, r in zip(samples, p(samples)):
        print(f"  {s!r} -> {r['label']} ({r['score']:.2f})")
```
